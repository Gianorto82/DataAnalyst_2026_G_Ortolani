```python
import re
import pandas as pd

def estrai_posizione_numerica(valore):
    testo = str(valore).strip()

    # Posizioni che iniziano direttamente con un numero:
    # 1º, 2º (pari merito), 6°, 2...
    risultato = re.match(r"^(\d+)", testo)

    if risultato:
        return int(risultato.group(1))

    # Casi come F (4º)[19]
    risultato_finalista = re.match(r"^F \((\d+)[º°]\)", testo)

    if risultato_finalista:
        return int(risultato_finalista.group(1))

    return pd.NA


classifica["posizione_numerica"] = (
    classifica["posizione"]
    .apply(estrai_posizione_numerica)
    .astype("Int64")
)


def crea_categoria_risultato(riga):
    posizione = riga["posizione_numerica"]
    testo = str(riga["posizione"]).strip()

    if pd.notna(posizione):
        if posizione == 1:
            return "Vincitore"
        elif posizione in [2, 3]:
            return "Podio"
        else:
            return "Altra posizione"

    elif testo == "F":
        return "Finalista senza posizione"

    elif testo.startswith("NF"):
        return "Non finalista"

    elif testo == "NC":
        return "Non classificata"

    elif testo.startswith("Esclus"):
        return "Esclusa/o"

    else:
        return "Altro"


classifica["categoria_risultato"] = classifica.apply(
    crea_categoria_risultato,
    axis=1
)

print("Conteggio per categoria:")
print(classifica["categoria_risultato"].value_counts().to_string())

print("\nEsempio delle nuove colonne:")
display(
    classifica[
        ["posizione", "posizione_numerica", "categoria_risultato"]
    ].sample(15, random_state=42)
)
```

```python
import unicodedata
import re

def normalizza_testo(valore):
    # Converte il valore in testo e lo rende minuscolo
    testo = str(valore).lower().strip()

    # Separa le lettere dagli accenti:
    # "perché" diventa temporaneamente "perché"
    testo = unicodedata.normalize("NFKD", testo)

    # Elimina accenti e altri segni combinati
    testo = "".join(
        carattere for carattere in testo
        if not unicodedata.combining(carattere)
    )

    # Sostituisce punteggiatura e simboli con uno spazio
    testo = re.sub(r"[^a-z0-9]+", " ", testo)

    # Elimina spazi multipli
    testo = re.sub(r"\s+", " ", testo).strip()

    return testo


# Chiavi nella classifica
classifica["chiave_canzone"] = classifica["canzone"].apply(normalizza_testo)
classifica["chiave_artista"] = classifica["interprete"].apply(normalizza_testo)

# Chiavi nei dati Spotify
spotify["chiave_canzone"] = spotify["canzone"].apply(normalizza_testo)
spotify["chiave_artista"] = spotify["artista"].apply(normalizza_testo)

print("Esempi dalla classifica:")
display(
    classifica[
        ["canzone", "chiave_canzone", "interprete", "chiave_artista"]
    ].head(10)
)

print("\nEsempi da Spotify:")
display(
    spotify[
        ["canzone", "chiave_canzone", "artista", "chiave_artista"]
    ].head(10)
)
```

```python
# Insieme delle coppie titolo-artista presenti su Spotify
coppie_spotify = set(
    zip(spotify["chiave_canzone"], spotify["chiave_artista"])
)

# Insieme dei titoli presenti su Spotify
titoli_spotify = set(spotify["chiave_canzone"])


# Verifica dell'abbinamento esatto titolo + artista
classifica["match_titolo_artista"] = [
    (titolo, artista) in coppie_spotify
    for titolo, artista in zip(
        classifica["chiave_canzone"],
        classifica["chiave_artista"]
    )
]

# Verifica dell'abbinamento basato soltanto sul titolo
classifica["match_solo_titolo"] = (
    classifica["chiave_canzone"].isin(titoli_spotify)
)


totale = len(classifica)
match_esatti = classifica["match_titolo_artista"].sum()
match_titolo = classifica["match_solo_titolo"].sum()

print(
    "Match titolo + artista:",
    match_esatti,
    f"({match_esatti / totale:.1%})"
)

print(
    "Match basati sul titolo:",
    match_titolo,
    f"({match_titolo / totale:.1%})"
)

print(
    "Titolo trovato ma artista differente:",
    (
        classifica["match_solo_titolo"]
        & ~classifica["match_titolo_artista"]
    ).sum()
)
```

```python
# Numero di righe Spotify associate a ogni titolo normalizzato
conteggio_titoli_spotify = (
    spotify["chiave_canzone"]
    .value_counts()
)

# Titoli presenti più di una volta su Spotify
titoli_ambigui = conteggio_titoli_spotify[
    conteggio_titoli_spotify > 1
]

print("Titoli Spotify distinti:", spotify["chiave_canzone"].nunique())
print("Titoli ambigui:", len(titoli_ambigui))
print("Righe Spotify coinvolte:", titoli_ambigui.sum())

print("\nTitoli con più corrispondenze:")
print(titoli_ambigui.head(20).to_string())

# Mostriamo alcuni casi completi
esempi_ambigui = (
    spotify[
        spotify["chiave_canzone"].isin(titoli_ambigui.index)
    ]
    .sort_values(["chiave_canzone", "popolarita"], ascending=[True, False])
)

display(
    esempi_ambigui[
        ["canzone", "artista", "anno_ultimo_rilascio", "popolarita"]
    ].head(30)
)
```

```python
# Conserviamo una sola riga Spotify per ogni titolo normalizzato
spotify_univoco = (
    spotify
    .drop_duplicates(subset=["chiave_canzone"], keep="first")
    .copy()
)

# Selezioniamo e rinominiamo i campi Spotify utili
spotify_per_unione = spotify_univoco[
    [
        "chiave_canzone",
        "canzone",
        "artista",
        "anno_ultimo_rilascio",
        "durata_min",
        "popolarita",
        "url_album"
    ]
].rename(columns={
    "canzone": "canzone_spotify",
    "artista": "artista_spotify"
})

# Unione: manteniamo tutte le righe della classifica
sanremo = classifica.merge(
    spotify_per_unione,
    on="chiave_canzone",
    how="left",
    validate="many_to_one"
)

# Indichiamo se i dati Spotify sono disponibili
sanremo["dati_spotify_disponibili"] = (
    sanremo["popolarita"].notna()
)

print("Righe prima dell'unione:", len(classifica))
print("Righe dopo l'unione:", len(sanremo))
print(
    "Righe con dati Spotify:",
    sanremo["dati_spotify_disponibili"].sum()
)
print(
    "Righe senza dati Spotify:",
    (~sanremo["dati_spotify_disponibili"]).sum()
)

display(
    sanremo[
        [
            "anno_festival",
            "canzone",
            "interprete",
            "categoria_risultato",
            "artista_spotify",
            "popolarita"
        ]
    ].head(10)
)
```

```python
# Uniamo le informazioni generali del Festival
sanremo = sanremo.merge(
    festival,
    on="anno_festival",
    how="left",
    validate="many_to_one"
)

print("Righe dopo l'unione con Festival:", len(sanremo))

print("\nValori mancanti nei campi del Festival:")
print(
    sanremo[
        [
            "periodo",
            "sede",
            "presentatore",
            "partecipanti",
            "vincitore"
        ]
    ].isna().sum()
)

display(
    sanremo[
        [
            "anno_festival",
            "canzone",
            "interprete",
            "posizione",
            "presentatore",
            "partecipanti",
            "vincitore"
        ]
    ].sample(10, random_state=42)
)
```

```python
# Impostiamo inizialmente tutte le righe come non tardive
sanremo["ristampa_tardiva"] = "No"

# Righe senza informazioni Spotify
sanremo.loc[
    sanremo["anno_ultimo_rilascio"].isna(),
    "ristampa_tardiva"
] = "Dati non disponibili"

# Righe con ultimo rilascio avvenuto almeno 20 anni dopo
sanremo.loc[
    sanremo["anni_tra_festival_e_rilascio"].ge(20).fillna(False),
    "ristampa_tardiva"
] = "Sì"


# Fasce di popolarità Spotify
sanremo["fascia_popolarita"] = pd.cut(
    sanremo["popolarita"],
    bins=[-1, 20, 40, 60, 100],
    labels=[
        "Bassa (0-20)",
        "Medio-bassa (21-40)",
        "Media (41-60)",
        "Alta (61-100)"
    ]
)

sanremo["fascia_popolarita"] = (
    sanremo["fascia_popolarita"]
    .astype("object")
    .fillna("Dati non disponibili")
)


print("Decenni:")
print(sanremo["decennio"].value_counts().sort_index().to_string())

print("\nRistampa tardiva:")
print(sanremo["ristampa_tardiva"].value_counts().to_string())

print("\nFasce di popolarità:")
print(sanremo["fascia_popolarita"].value_counts().to_string())
```

```python
print("Statistiche sulla distanza temporale:")
print(
    sanremo["anni_tra_festival_e_rilascio"]
    .describe()
)

rilasci_precedenti = sanremo[
    sanremo["anni_tra_festival_e_rilascio"] < 0
].copy()

print(
    "\nRighe con ultimo rilascio precedente al Festival:",
    len(rilasci_precedenti)
)

display(
    rilasci_precedenti[
        [
            "anno_festival",
            "canzone",
            "interprete",
            "anno_ultimo_rilascio",
            "artista_spotify",
            "popolarita",
            "anni_tra_festival_e_rilascio"
        ]
    ].sort_values(
        "anni_tra_festival_e_rilascio"
    )
)
```

```python
# Normalizziamo anche il nome dell'artista arrivato da Spotify
sanremo["chiave_artista_spotify"] = (
    sanremo["artista_spotify"]
    .fillna("")
    .apply(normalizza_testo)
)


def verifica_artista_compatibile(riga):
    artista_sanremo = riga["chiave_artista"]
    artista_spotify = riga["chiave_artista_spotify"]

    if artista_spotify == "":
        return False

    # Coincidenza esatta oppure nome contenuto nell'altro:
    # utile per casi con due o più interpreti
    return (
        artista_sanremo == artista_spotify
        or artista_spotify in artista_sanremo
        or artista_sanremo in artista_spotify
    )


sanremo["artista_compatibile"] = sanremo.apply(
    verifica_artista_compatibile,
    axis=1
)

# Accettiamo un rilascio nello stesso anno, successivo
# oppure al massimo un anno precedente al Festival
sanremo["anno_coerente"] = (
    sanremo["anni_tra_festival_e_rilascio"]
    .ge(-1)
    .fillna(False)
    .astype(bool)
)

# Un abbinamento affidabile deve rispettare entrambe le condizioni
sanremo["abbinamento_affidabile"] = (
    sanremo["artista_compatibile"]
    & sanremo["anno_coerente"]
)


# Creiamo un'etichetta descrittiva
sanremo["qualita_abbinamento"] = "Da verificare: artista differente"

sanremo.loc[
    ~sanremo["dati_spotify_disponibili"],
    "qualita_abbinamento"
] = "Senza dati Spotify"

sanremo.loc[
    sanremo["anni_tra_festival_e_rilascio"].lt(-1).fillna(False),
    "qualita_abbinamento"
] = "Da escludere: rilascio precedente"

sanremo.loc[
    sanremo["abbinamento_affidabile"],
    "qualita_abbinamento"
] = "Affidabile"


print(
    sanremo["qualita_abbinamento"]
    .value_counts()
    .to_string()
)
```

```python
# Campo semplice da utilizzare come filtro in Looker Studio
sanremo["includi_analisi_spotify"] = "No"

sanremo.loc[
    sanremo["abbinamento_affidabile"],
    "includi_analisi_spotify"
] = "Sì"


# Ricalcoliamo la ristampa tardiva soltanto sui match affidabili
sanremo["ristampa_tardiva"] = "Non valutabile"

sanremo.loc[
    sanremo["abbinamento_affidabile"]
    & sanremo["anni_tra_festival_e_rilascio"].lt(20).fillna(False),
    "ristampa_tardiva"
] = "No"

sanremo.loc[
    sanremo["abbinamento_affidabile"]
    & sanremo["anni_tra_festival_e_rilascio"].ge(20).fillna(False),
    "ristampa_tardiva"
] = "Sì"


print("Righe utilizzabili per l'analisi Spotify:")
print(
    sanremo["includi_analisi_spotify"]
    .value_counts()
    .to_string()
)

print("\nMatch affidabili per decennio:")
print(
    sanremo[
        sanremo["includi_analisi_spotify"] == "Sì"
    ]["decennio"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nMatch affidabili per risultato:")
print(
    sanremo[
        sanremo["includi_analisi_spotify"] == "Sì"
    ]["categoria_risultato"]
    .value_counts()
    .to_string()
)

print("\nRistampa tardiva:")
print(
    sanremo["ristampa_tardiva"]
    .value_counts()
    .to_string()
)
```

```python
from difflib import SequenceMatcher

def calcola_somiglianza_artisti(riga):
    artista_sanremo = riga["chiave_artista"]
    artista_spotify = riga["chiave_artista_spotify"]

    if artista_spotify == "":
        return pd.NA

    return round(
        SequenceMatcher(
            None,
            artista_sanremo,
            artista_spotify
        ).ratio(),
        3
    )


sanremo["somiglianza_artisti"] = sanremo.apply(
    calcola_somiglianza_artisti,
    axis=1
).astype("Float64")


# Esaminiamo solo i casi:
# - con dati Spotify;
# - con anno temporalmente coerente;
# - non già considerati affidabili.
candidati_da_rivedere = sanremo[
    sanremo["dati_spotify_disponibili"]
    & sanremo["anno_coerente"]
    & ~sanremo["artista_compatibile"]
].copy()


candidati_da_rivedere["fascia_somiglianza"] = pd.cut(
    candidati_da_rivedere["somiglianza_artisti"],
    bins=[0, 0.30, 0.50, 0.70, 0.85, 1],
    labels=[
        "Molto bassa (0-0,30)",
        "Bassa (0,31-0,50)",
        "Media (0,51-0,70)",
        "Alta (0,71-0,85)",
        "Molto alta (oltre 0,85)"
    ],
    include_lowest=True
)


print("Candidati da rivedere:", len(candidati_da_rivedere))

print("\nDistribuzione della somiglianza:")
print(
    candidati_da_rivedere["fascia_somiglianza"]
    .value_counts()
    .sort_index()
    .to_string()
)

print("\nCasi con maggiore somiglianza:")
display(
    candidati_da_rivedere[
        [
            "anno_festival",
            "canzone",
            "interprete",
            "artista_spotify",
            "anno_ultimo_rilascio",
            "somiglianza_artisti"
        ]
    ]
    .sort_values("somiglianza_artisti", ascending=False)
    .head(30)
)
```

```python
# Parole troppo generiche per identificare un artista
parole_generiche = {
    "e", "con", "and", "the", "i", "il", "la", "le", "gli",
    "di", "dei", "del", "della", "band", "gruppo", "orchestra"
}


def condivide_nome_distintivo(riga):
    parole_sanremo = set(
        riga["chiave_artista"].split()
    ) - parole_generiche

    parole_spotify = set(
        riga["chiave_artista_spotify"].split()
    ) - parole_generiche

    parole_comuni = parole_sanremo.intersection(parole_spotify)

    # Accettiamo solo parole comuni abbastanza lunghe,
    # come "Rettore", "Facchinetti" o "Camaleonti"
    return any(
        len(parola) >= 6
        for parola in parole_comuni
    )


sanremo["nome_distintivo_comune"] = sanremo.apply(
    condivide_nome_distintivo,
    axis=1
)


# Nuovo criterio proposto
sanremo["abbinamento_affidabile_v2"] = (
    sanremo["dati_spotify_disponibili"]
    & sanremo["anno_coerente"]
    & (
        sanremo["artista_compatibile"]
        | sanremo["somiglianza_artisti"].ge(0.70).fillna(False)
        | (
            sanremo["nome_distintivo_comune"]
            & sanremo["somiglianza_artisti"].ge(0.45).fillna(False)
        )
    )
)


recuperati = sanremo[
    sanremo["abbinamento_affidabile_v2"]
    & ~sanremo["abbinamento_affidabile"]
].copy()


print(
    "Abbinamenti affidabili con il vecchio criterio:",
    sanremo["abbinamento_affidabile"].sum()
)

print(
    "Abbinamenti affidabili con il nuovo criterio:",
    sanremo["abbinamento_affidabile_v2"].sum()
)

print(
    "Nuovi abbinamenti recuperati:",
    len(recuperati)
)

display(
    recuperati[
        [
            "anno_festival",
            "canzone",
            "interprete",
            "artista_spotify",
            "anno_ultimo_rilascio",
            "somiglianza_artisti",
            "nome_distintivo_comune"
        ]
    ].sort_values(
        "somiglianza_artisti",
        ascending=False
    )
)
```

```python
# Sostituiamo il vecchio criterio con quello migliorato
sanremo["abbinamento_affidabile"] = (
    sanremo["abbinamento_affidabile_v2"]
)

# Aggiorniamo il campo destinato al filtro di Looker Studio
sanremo["includi_analisi_spotify"] = "No"

sanremo.loc[
    sanremo["abbinamento_affidabile"],
    "includi_analisi_spotify"
] = "Sì"


# Aggiorniamo la qualità dell'abbinamento
sanremo["qualita_abbinamento"] = "Da verificare: artista differente"

sanremo.loc[
    ~sanremo["dati_spotify_disponibili"],
    "qualita_abbinamento"
] = "Senza dati Spotify"

sanremo.loc[
    sanremo["anni_tra_festival_e_rilascio"].lt(-1).fillna(False),
    "qualita_abbinamento"
] = "Da escludere: rilascio precedente"

sanremo.loc[
    sanremo["abbinamento_affidabile"],
    "qualita_abbinamento"
] = "Affidabile"


# Aggiorniamo la classificazione delle ristampe
sanremo["ristampa_tardiva"] = "Non valutabile"

sanremo.loc[
    sanremo["abbinamento_affidabile"]
    & sanremo["anni_tra_festival_e_rilascio"].lt(20).fillna(False),
    "ristampa_tardiva"
] = "No"

sanremo.loc[
    sanremo["abbinamento_affidabile"]
    & sanremo["anni_tra_festival_e_rilascio"].ge(20).fillna(False),
    "ristampa_tardiva"
] = "Sì"


print("Qualità finale degli abbinamenti:")
print(
    sanremo["qualita_abbinamento"]
    .value_counts()
    .to_string()
)

print("\nRighe incluse nell'analisi Spotify:")
print(
    sanremo["includi_analisi_spotify"]
    .value_counts()
    .to_string()
)

print("\nRistampe tardive:")
print(
    sanremo["ristampa_tardiva"]
    .value_counts()
    .to_string()
)
```

##1: Esiste una relazione tra la posizione raggiunta a Sanremo e la popolarità attuale?

```python
# Selezioniamo soltanto:
# - abbinamenti Spotify affidabili;
# - canzoni con una posizione numerica
analisi_posizione = sanremo[
    sanremo["abbinamento_affidabile"]
    & sanremo["posizione_numerica"].notna()
].copy()


# Creiamo fasce di posizione
analisi_posizione["fascia_posizione"] = pd.cut(
    analisi_posizione["posizione_numerica"],
    bins=[0, 1, 3, 10, 100],
    labels=[
        "1º posto",
        "2º-3º posto",
        "4º-10º posto",
        "Oltre il 10º posto"
    ]
)


# Riepilogo della popolarità per fascia
riepilogo_posizione = (
    analisi_posizione
    .groupby("fascia_posizione", observed=True)
    .agg(
        numero_canzoni=("canzone", "count"),
        popolarita_media=("popolarita", "mean"),
        popolarita_mediana=("popolarita", "median"),
        popolarita_minima=("popolarita", "min"),
        popolarita_massima=("popolarita", "max")
    )
    .round(2)
    .reset_index()
)


# Correlazione per ranghi
correlazione = (
    analisi_posizione["posizione_numerica"]
    .rank()
    .corr(
        analisi_posizione["popolarita"].rank()
    )
)


print(
    "Canzoni utilizzate nell'analisi:",
    len(analisi_posizione)
)

print(
    "Correlazione tra posizione e popolarità:",
    round(correlazione, 3)
)

display(riepilogo_posizione)
```

Esiste una relazione negativa ma debole tra posizione e popolarità (-0,174): le canzoni arrivate nelle posizioni migliori tendono a essere oggi leggermente più popolari, ma il risultato a Sanremo non è sufficiente, da solo, a spiegare il successo su Spotify.


```python
analisi_decennio = sanremo[
    sanremo["abbinamento_affidabile"]
].copy()


riepilogo_decennio = (
    analisi_decennio
    .groupby(
        ["decennio_num", "decennio"],
        as_index=False
    )
    .agg(
        numero_canzoni=("canzone", "count"),
        popolarita_media=("popolarita", "mean"),
        popolarita_mediana=("popolarita", "median"),
        popolarita_massima=("popolarita", "max"),
        durata_media=("durata_min", "mean")
    )
    .sort_values("decennio_num")
)


riepilogo_decennio[
    [
        "popolarita_media",
        "popolarita_mediana",
        "durata_media"
    ]
] = riepilogo_decennio[
    [
        "popolarita_media",
        "popolarita_mediana",
        "durata_media"
    ]
].round(2)


display(riepilogo_decennio)
```

Il risultato mostra un forte effetto generazionale:


*   gli anni 2020 hanno la popolarità media più alta: 46,72;
*   seguono anni 2010 (31,67) e anni 1980 (31,11);
*   gli anni 1950 hanno una media molto bassa (5,68);
*   gli anni 1980 superano sia anni 1990 sia anni 2000.


La mediana conferma il risultato, quindi non dipende soltanto da pochi grandi successi. Tuttavia, i brani più recenti sono naturalmente favoriti sulle piattaforme digitali: non possiamo dire che siano “più resistenti”, perché hanno avuto meno tempo per invecchiare.

Anche la durata racconta qualcosa: raggiunge il massimo negli anni 1990 (4,25 minuti) e scende a 3,40 negli anni 2020.

```python
# Consideriamo soltanto canzoni storiche con match affidabile
analisi_ristampe = sanremo[
    sanremo["abbinamento_affidabile"]
    & (sanremo["anno_festival"] <= 2003)
].copy()


# Ultimo rilascio avvenuto dal 2015 in avanti
analisi_ristampe["rilascio_recente"] = "No"

analisi_ristampe.loc[
    analisi_ristampe["anno_ultimo_rilascio"] >= 2015,
    "rilascio_recente"
] = "Sì"


riepilogo_ristampe = (
    analisi_ristampe
    .groupby("rilascio_recente", as_index=False)
    .agg(
        numero_canzoni=("canzone", "count"),
        popolarita_media=("popolarita", "mean"),
        popolarita_mediana=("popolarita", "median"),
        popolarita_massima=("popolarita", "max")
    )
)


# Percentuale di brani con popolarità almeno pari a 40
percentuale_popolari = (
    analisi_ristampe
    .assign(brano_popolare=analisi_ristampe["popolarita"] >= 40)
    .groupby("rilascio_recente")["brano_popolare"]
    .mean()
    .mul(100)
    .round(1)
    .rename("percentuale_con_popolarita_40_o_piu")
    .reset_index()
)


riepilogo_ristampe = riepilogo_ristampe.merge(
    percentuale_popolari,
    on="rilascio_recente"
)

riepilogo_ristampe[
    ["popolarita_media", "popolarita_mediana"]
] = riepilogo_ristampe[
    ["popolarita_media", "popolarita_mediana"]
].round(2)


print(
    "Canzoni storiche analizzate:",
    len(analisi_ristampe)
)

display(riepilogo_ristampe)
```

Il risultato è netto e contrario all’ipotesi iniziale:


*   brani con rilascio precedente al 2015: popolarità media 24,89;
*   brani ripubblicati dal 2015: media 13,15;
*   la percentuale con popolarità almeno 40 scende dal 22,5% al 3,4%.


Le ristampe recenti non risultano associate a una maggiore popolarità. Al contrario, i brani storici con ultimo rilascio recente mostrano valori medi più bassi.

Una possibile interpretazione è che molte ristampe appartengano a raccolte digitali di brani poco conosciuti. Inoltre, l’“anno dell’ultimo rilascio” fornito da Spotify non dimostra necessariamente una campagna promozionale.

```python
risultati_soglie = []

for soglia in [2010, 2015, 2020]:

    temporaneo = analisi_ristampe.copy()

    temporaneo["gruppo_rilascio"] = (
        temporaneo["anno_ultimo_rilascio"] >= soglia
    ).map({
        True: f"Dal {soglia}",
        False: f"Prima del {soglia}"
    })

    riepilogo = (
        temporaneo
        .groupby("gruppo_rilascio")
        .agg(
            numero_canzoni=("canzone", "count"),
            popolarita_media=("popolarita", "mean"),
            popolarita_mediana=("popolarita", "median")
        )
        .round(2)
        .reset_index()
    )

    riepilogo["soglia"] = soglia
    risultati_soglie.append(riepilogo)


sensibilita_ristampe = pd.concat(
    risultati_soglie,
    ignore_index=True
)

display(
    sensibilita_ristampe[
        [
            "soglia",
            "gruppo_rilascio",
            "numero_canzoni",
            "popolarita_media",
            "popolarita_mediana"
        ]
    ].sort_values(["soglia", "gruppo_rilascio"])
)
```

Nel campione analizzato, una ristampa recente non è associata a una popolarità più elevata. Il risultato rimane stabile usando soglie temporali differenti.

```python
canzoni_storiche = (
    sanremo[
        sanremo["abbinamento_affidabile"]
        & (sanremo["anno_festival"] <= 2003)
    ]
    .sort_values(
        ["popolarita", "anno_festival"],
        ascending=[False, True]
    )
    .drop_duplicates(
        subset=["chiave_canzone"],
        keep="first"
    )
    .copy()
)


# Età della canzone rispetto all'ultima edizione del dataset
canzoni_storiche["anni_dal_festival"] = (
    2023 - canzoni_storiche["anno_festival"]
)


top_canzoni_storiche = canzoni_storiche[
    [
        "anno_festival",
        "canzone",
        "interprete",
        "posizione",
        "popolarita",
        "anno_ultimo_rilascio",
        "anni_dal_festival",
        "url_album"
    ]
].head(15)


display(top_canzoni_storiche)
```

Il risultato ottenuto a Sanremo è associato solo debolmente alla popolarità futura. Alcuni brani inizialmente classificati male sono diventati successi duraturi.

```python
analisi_durata = sanremo[
    sanremo["abbinamento_affidabile"]
].copy()


print("Statistiche generali sulla durata:")
print(
    analisi_durata["durata_min"]
    .describe()
    .round(2)
)


durata_decennio = (
    analisi_durata
    .groupby(
        ["decennio_num", "decennio"],
        as_index=False
    )
    .agg(
        numero_canzoni=("canzone", "count"),
        durata_media=("durata_min", "mean"),
        durata_mediana=("durata_min", "median"),
        durata_minima=("durata_min", "min"),
        durata_massima=("durata_min", "max")
    )
    .sort_values("decennio_num")
)


durata_decennio[
    [
        "durata_media",
        "durata_mediana",
        "durata_minima",
        "durata_massima"
    ]
] = durata_decennio[
    [
        "durata_media",
        "durata_mediana",
        "durata_minima",
        "durata_massima"
    ]
].round(2)


display(durata_decennio)
```

La durata non è cambiata in modo lineare. Le canzoni si sono progressivamente allungate tra gli anni Sessanta e Novanta, per poi accorciarsi nuovamente nell’era digitale.

```python
# Una sola riga per ogni canzone storica
storiche_uniche = (
    sanremo[
        sanremo["abbinamento_affidabile"]
        & (sanremo["anno_festival"] <= 2003)
    ]
    .sort_values("popolarita", ascending=False)
    .drop_duplicates(
        subset=["chiave_canzone"],
        keep="first"
    )
    .copy()
)


riepilogo_artisti = (
    storiche_uniche
    .groupby("artista_spotify", as_index=False)
    .agg(
        numero_canzoni=("chiave_canzone", "nunique"),
        popolarita_media=("popolarita", "mean"),
        popolarita_mediana=("popolarita", "median"),
        popolarita_massima=("popolarita", "max")
    )
)


# Manteniamo artisti con almeno due canzoni
top_artisti_storici = (
    riepilogo_artisti[
        riepilogo_artisti["numero_canzoni"] >= 2
    ]
    .sort_values(
        ["popolarita_media", "numero_canzoni"],
        ascending=[False, False]
    )
    .head(15)
    .copy()
)


top_artisti_storici[
    ["popolarita_media", "popolarita_mediana"]
] = top_artisti_storici[
    ["popolarita_media", "popolarita_mediana"]
].round(2)


display(top_artisti_storici)
```

```python
# Fascia della posizione direttamente nel dataset principale
sanremo["fascia_posizione"] = pd.cut(
    sanremo["posizione_numerica"],
    bins=[0, 1, 3, 10, 100],
    labels=[
        "1º posto",
        "2º-3º posto",
        "4º-10º posto",
        "Oltre il 10º posto"
    ]
)

sanremo["fascia_posizione"] = (
    sanremo["fascia_posizione"]
    .astype("object")
    .fillna("Senza posizione numerica")
)


# Età della canzone al termine del periodo analizzato
sanremo["anni_dal_festival"] = (
    2023 - sanremo["anno_festival"]
)


# Indicazione di canzone storica
sanremo["canzone_storica"] = "No"

sanremo.loc[
    sanremo["anno_festival"] <= 2003,
    "canzone_storica"
] = "Sì"


# Rilascio recente utilizzabile soltanto per i brani storici affidabili
sanremo["rilascio_recente_2015"] = "Non valutabile"

condizione_storica_affidabile = (
    sanremo["abbinamento_affidabile"]
    & (sanremo["anno_festival"] <= 2003)
)

sanremo.loc[
    condizione_storica_affidabile
    & (sanremo["anno_ultimo_rilascio"] < 2015),
    "rilascio_recente_2015"
] = "No"

sanremo.loc[
    condizione_storica_affidabile
    & (sanremo["anno_ultimo_rilascio"] >= 2015),
    "rilascio_recente_2015"
] = "Sì"


# Identifichiamo una sola riga per ogni titolo Spotify affidabile
sanremo["record_canzone_unico"] = "No"

indici_unici = (
    sanremo[sanremo["abbinamento_affidabile"]]
    .sort_values(
        ["popolarita", "anno_festival"],
        ascending=[False, True]
    )
    .drop_duplicates(
        subset=["chiave_canzone"],
        keep="first"
    )
    .index
)

sanremo.loc[
    indici_unici,
    "record_canzone_unico"
] = "Sì"


# Selezioniamo soltanto i campi utili per il report
colonne_looker = [
    "anno_festival",
    "decennio_num",
    "decennio",
    "canzone",
    "interprete",
    "autori",
    "posizione",
    "posizione_numerica",
    "fascia_posizione",
    "categoria_risultato",
    "periodo",
    "sede",
    "presentatore",
    "partecipanti",
    "vincitore",
    "artista_spotify",
    "anno_ultimo_rilascio",
    "durata_min",
    "popolarita",
    "fascia_popolarita",
    "url_album",
    "anni_tra_festival_e_rilascio",
    "anni_dal_festival",
    "canzone_storica",
    "ristampa_tardiva",
    "rilascio_recente_2015",
    "qualita_abbinamento",
    "includi_analisi_spotify",
    "record_canzone_unico"
]

dataset_looker = sanremo[colonne_looker].copy()


print("Dimensioni dataset finale:", dataset_looker.shape)

print("\nConteggio record unici affidabili:")
print(
    dataset_looker["record_canzone_unico"]
    .value_counts()
    .to_string()
)

print("\nPrime cinque righe:")
display(dataset_looker.head())
```
